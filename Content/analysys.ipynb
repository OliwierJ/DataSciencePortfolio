## Data Science Portfolio - Game Meta Data Analysis

This section of the portfolio focuses on simple analysis of chess game results and meta data (e.g. game start time, move count)


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('../Resources/games_data.csv', index_col=0)
data.head()
```

```python
# Plot the distribution of game start times
X = pd.to_datetime(data['created_at'], unit='ms')

plt.figure(figsize=(12, 6))
plt.hist(X, bins=60, color='blue', alpha=0.7, log=False)
plt.title('Distribution of Game Start Times')
plt.xlabel('Game Start Time')
plt.ylabel('Number of Games')
```
